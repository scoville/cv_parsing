# Measure average time and cost for running ocr on one document

```python
from azure.core.credentials import AzureKeyCredential
from azure.core.exceptions import AzureError
import os
import pprint
from azure.ai.documentintelligence import DocumentIntelligenceClient
```

```python
# AZURE_CONTENT_UNDERSTANDING_ENDPOINT - the endpoint to your Content Understanding resource.
endpoint = os.environ["CONTENTUNDERSTANDING_ENDPOINT"]
# CONTENT_UNDERSTANDING_KEY - your Content Understanding API key
key = os.environ["CONTENTUNDERSTANDING_KEY"]
credential = AzureKeyCredential(key)
```

```python
# local file localtion
path_to_sample_document = "./examples/pdf/職務経歴書(サンプル).pdf"
```

```python
def run_ocr_only(path_to_sample_document: str, client: DocumentIntelligenceClient):
    with open(path_to_sample_document, "rb") as f:
        poller = client.begin_analyze_document(
            "prebuilt-read",
            body=f,
            output_content_format="markdown",
        )
    
    result = poller.result()
    return result, poller
```

```python
# Set up Document Intelligence client.
client = DocumentIntelligenceClient(endpoint=endpoint, credential=credential)
```

```python
result, poller = run_ocr_only(path_to_sample_document, client)
```

```python
%%timeit -r 7 -n 1 
# Calulate average time for only running ocr on a document (2-page pdf)
result, poller = run_ocr_only(path_to_sample_document, client)
```

```python
# ocr result
markdown = result.content
print(markdown)
```

```python
price= 1.5 /1000 *2
print(f"cost (usd) per document(2-page pdf) for running ocr only: {price}")
```
